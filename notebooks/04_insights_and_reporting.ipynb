# Ansiedad en estudiantes de ingeniería en sistemas

## Reporte final basado en el Inventario de Ansiedad de Beck (BAI)

Este reporte sintetiza el análisis realizado sobre una muestra de estudiantes de ingeniería en sistemas. A partir de las respuestas del BAI, el proyecto explora la distribución de la ansiedad, los síntomas más relevantes, la presencia de perfiles diferenciados y la capacidad de un modelo supervisado para estimar niveles de ansiedad.

La lectura debe entenderse como un análisis descriptivo y exploratorio. Los resultados ayudan a identificar patrones, pero no sustituyen evaluación clínica ni permiten inferir causalidad.


## Panorama del dataset

- Registros analizados: **113 estudiantes**.
- Instrumento: **21 ítems BAI**, cada uno representado como `BAI_1` a `BAI_21`.
- Score principal: **`totalScore`**, calculado como la suma de los 21 ítems.
- Niveles de ansiedad: **LOW**, **MODERATE** y **SEVERE**.

Distribución observada:

| Categoría | Casos | Lectura |
|---|---:|---|
| LOW | 71 | Mayor parte de la muestra |
| MODERATE | 34 | Grupo relevante para seguimiento temprano |
| SEVERE | 8 | Grupo pequeño, pero de alta prioridad interpretativa |

En conjunto, **37.2% de la muestra** se ubica en niveles moderados o severos.


## Ruta metodológica

El análisis se desarrolló en cuatro etapas:

1. **Limpieza y preparación**: expansión de respuestas BAI, estandarización de categorías, validación de columnas y remoción de campos sensibles.
2. **Análisis exploratorio**: distribuciones, síntomas principales, diferencias por grupos y correlaciones.
3. **Modelado no supervisado**: KMeans para identificar perfiles de síntomas, con clusters ordenados por score promedio.
4. **Modelado supervisado**: Random Forest para estimar la categoría de ansiedad usando solo los síntomas BAI.

Una decisión metodológica importante fue no usar `totalScore` como predictor del modelo supervisado, porque es la suma directa de los ítems y filtraría la respuesta al clasificador.


# Hallazgos clave

## 1. La ansiedad moderada o severa no es marginal

Aunque la categoría `LOW` concentra la mayoría de los casos, más de un tercio de la muestra cae en `MODERATE` o `SEVERE`. Esto convierte la ansiedad relevante en un patrón de interés dentro del grupo, no en una excepción aislada.

La mediana del `totalScore` es 16, pero el rango llega hasta 61, lo que muestra una distribución con casos de severidad alta que merecen interpretación cuidadosa.


## 2. La severidad se expresa con fuerza en síntomas físicos

Los síntomas más informativos para el modelo final incluyen:

- Inestabilidad.
- Temblor general.
- Miedo a perder el control.
- Dificultad para respirar.
- Temblor en las manos.
- Incapacidad para relajarse.
- Terror o miedo.
- Latidos fuertes o rápidos.

La lectura principal es que, en esta muestra, la ansiedad aparece muy ligada a activación corporal y sensaciones fisiológicas, no solamente a preocupación abstracta o malestar general.


## 3. Los perfiles de ansiedad son diferenciables

El clustering con KMeans identifica tres perfiles ordenados por score promedio:

| Perfil | n | % | Score promedio | Interpretación |
|---|---:|---:|---:|---|
| Cluster 1 | 62 | 54.9% | 10.63 | Baja ansiedad |
| Cluster 2 | 31 | 27.4% | 23.71 | Ansiedad intermedia |
| Cluster 3 | 20 | 17.7% | 35.70 | Mayor severidad |

El Cluster 1 queda compuesto por casos `LOW`. El Cluster 2 se concentra mayormente en `MODERATE`, mientras que el Cluster 3 agrupa los casos más altos e incluye el total de casos `SEVERE` junto con estudiantes moderados de score elevado.

Esto sugiere que el score total no solo sube de forma continua: también aparecen perfiles relativamente interpretables de síntomas.


## 4. El modelo predictivo captura señales útiles, pero con cautela

El Random Forest alcanza aproximadamente **79% de accuracy** en el conjunto de test. El desempeño es mejor para `LOW`, razonable para `MODERATE` y limitado para `SEVERE` por el bajo número de casos.

Esta diferencia es importante: el resultado demuestra que los síntomas contienen información predictiva, pero no convierte el modelo en una herramienta diagnóstica. Para uso institucional o clínico haría falta más muestra, validación externa y evaluación con criterios profesionales.


# Interpretación

La ansiedad observada en la muestra parece tener una estructura clara: niveles bajos para la mayoría, un grupo intermedio que conviene monitorear y un grupo más pequeño con scores altos.

Los síntomas fisiológicos aparecen como una vía central de manifestación. Esto puede ser útil para pensar en estrategias de detección temprana, especialmente cuando estudiantes no verbalizan la ansiedad como preocupación, pero sí reportan señales corporales intensas.

También es relevante que el grupo `SEVERE` sea pequeño. En términos estadísticos limita la estabilidad de los modelos; en términos prácticos, precisamente por su severidad, no debería descartarse ni promediarse hasta desaparecer.


# Outliers y casos extremos

Los valores extremos se conservaron porque, en datos psicológicos, un score alto no necesariamente representa un error. Puede representar un caso clínicamente relevante o una experiencia de ansiedad intensa.

Eliminar estos casos podría hacer que el análisis se vea más estable, pero también podría subestimar la severidad real presente en la muestra. Por eso se interpretan con cautela en lugar de descartarse automáticamente.


# Limitaciones

- Los datos son auto-reportados y pueden contener sesgos de respuesta.
- La muestra es relativamente pequeña.
- La clase `SEVERE` tiene pocos casos, lo que limita la evaluación del modelo.
- La muestra puede no representar a toda la población estudiantil.
- El diseño es transversal, por lo que no permite inferir causalidad.
- KMeans usa `k = 3` por interpretabilidad con las tres categorías BAI; no debe leerse como la única segmentación posible.

Estas limitaciones no invalidan el análisis, pero sí definen el nivel correcto de confianza.


# Implicaciones prácticas

Los resultados apuntan a tres ideas accionables:

- La detección temprana debería prestar atención a síntomas fisiológicos, no solo a reportes generales de preocupación.
- El grupo moderado merece seguimiento porque puede representar una zona de escalamiento antes de llegar a severidad alta.
- Los casos severos deben revisarse con prioridad y con apoyo profesional, no solo como observaciones extremas del dataset.

Para una institución educativa, el valor del análisis está en orientar preguntas, priorizar seguimiento y diseñar intervenciones más informadas.


# Conclusión

El proyecto muestra que la ansiedad en esta muestra es frecuente, medible y estructurada. A través del BAI, el análisis permite pasar de respuestas individuales a una lectura agregada de síntomas, niveles y perfiles.

La evidencia más fuerte está en la prevalencia de casos moderados o severos, la relevancia de síntomas físicos y la presencia de clusters interpretables. La siguiente mejora natural sería ampliar la muestra, validar los hallazgos con nuevos datos y complementar el análisis con variables académicas o contextuales.
